# In-Class Assignment 1


```python
# Income Statement Program

# Get sales input from user
sales = float(input("Enter sales ($): "))

# Calculate Cost of Goods Sold (40% of Sales)
cost_of_goods_sold = sales * 0.40

# Calculate Gross Profit
gross_profit = sales - cost_of_goods_sold

# Get tax rate from user
tax_rate = float(input("Enter tax rate (%): "))

# Calculate Taxes and Net Income
taxes = gross_profit * (tax_rate / 100)
net_income = gross_profit - taxes

# Display the Income Statement
print("\n Income Statement ")
print(f"Sales:              ${sales:,.2f}")
print(f"Cost of Goods Sold: ${cost_of_goods_sold:,.2f}")
print(f"Gross Profit:       ${gross_profit:,.2f}")
print(f"Taxes:              ${taxes:,.2f}")
print(f"Net Income:         ${net_income:,.2f}")
```

# In-Class Assignment 2

```python
# Simple Numerical Calculator

# Get user inputs
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operator = input("Enter an operator (+, -, *, /): ")

# Perform calculation
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        print("Error: Division by zero is not allowed.")
        exit()
else:
    print("Invalid operator.")
    exit()

# Print result
print("Result:", result)
```

```python
import math

ticket_price = 10
base_attendees = 20
fixed_cost = 200

print("Ad$   Profit")

for ad in range(0, 201, 25):
    # Calculate additional attendees
    additional = 2 * math.sqrt(ad)

    # Total attendees
    total_attendees = base_attendees + additional

    # Revenue
    revenue = ticket_price * total_attendees

    # Profit
    profit = revenue - ad - fixed_cost

    # Print results
    print(f"{ad:.1f}  {profit:.1f}")
```

```python
base_fee = 5
hourly_rate = 2.50
min_fee = 10
max_fee = 20

hours = 1

while hours <= 8:
    fee = base_fee + hourly_rate * hours

    # Apply minimum and maximum limits
    if fee < min_fee:
        fee = min_fee
    elif fee > max_fee:
        fee = max_fee

    print(hours, fee)

    hours += 1
```

# In-Class Assignment 4

```python
import random   # MUST be at the top

play = input("Do you want to play Rock-Paper-Scissors? (yes/no): ").lower()

while play == "yes":

    user_move = input("Choose rock, paper, or scissors: ").lower()

    # Random computer choice (this IS randomized)
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move == computer_move:
        print("It's a tie!")

    elif (user_move == "rock" and computer_move == "scissors") or \
         (user_move == "paper" and computer_move == "rock") or \
         (user_move == "scissors" and computer_move == "paper"):
        print("You win!")

    elif user_move in ["rock", "paper", "scissors"]:
        print("Computer wins!")

    else:
        print("Invalid choice.")

    play = input("Do you want to play again? (yes/no): ").lower()

print("Thanks for playing!")
```

```python
def reverse(s):
    # base case
    if len(s) <= 1:
        return s

    # recursive case
    return reverse(s[1:]) + s[0]

print(reverse('hello'))
print(reverse('12345'))
```

# In-Class Assignment 8

```python
def roi(cost, revenue):
    assert isinstance(cost, (int, float))
    assert isinstance(revenue, (int, float))
    assert cost != 0
    return (revenue - cost) / cost

def loan_payment(principal, annual_rate, months):
    assert isinstance(principal, (int, float))
    assert isinstance(annual_rate, (int, float))
    assert isinstance(months, (int, float))
    if principal <= 0:
        raise ValueError("principal must be positive")
    if months <= 0:
        raise ValueError("months must be positive")
    if annual_rate < 0:
        raise ValueError("annual_rate cannot be negative")
    r = annual_rate / 12
    if r != 0:
        pmt = r * principal / (1 - (1 + r) ** (-months))
    else:
        pmt = principal / months
    return pmt


try:
    cost = float(input("Enter cost: "))
    revenue = float(input("Enter revenue: "))
    print(roi(cost, revenue))
except Exception as e:
    print(e)


try:
    principal = float(input("Enter principal: "))
    annual_rate = float(input("Enter annual rate (decimal): "))
    months = int(input("Enter months: "))
    print(round(loan_payment(principal, annual_rate, months), 2))
except Exception as e:
    print(e)
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)


class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4

        # compute width and height when created
        self.width = self.calculate_width()
        self.height = self.calculate_height()

    def calculate_width(self):
        # assume p1 and p2 are width endpoints
        return self.p1.distance(self.p2)

    def calculate_height(self):
        # assume p2 and p3 are height endpoints
        return self.p2.distance(self.p3)

    def area(self):
        return self.width * self.height
```
